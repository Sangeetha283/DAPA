# Weather Data Analysis Using Python

This project analyzes weather data using pandas, matplotlib, and seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seab as sns
sns.set(style='whitegrid')
```

```python
# Load dataset
df = pd.read_csv(r"C:/Users/Admin/Downloads/Weather Data.csv")

# Convert date/time
df['Date/Time'] = pd.to_datetime(df['Date/Time'])
df['Month'] = df['Date/Time'].dt.month
df['Hour'] = df['Date/Time'].dt.hour
df['Date'] = df['Date/Time'].dt.date
```

```python
# Dataset Overview
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print(df.head())
print(df.describe())
```

```python
monthly_temp = df.groupby('Month')['Temp_C'].mean()

plt.figure(figsize=(10,5))
sns.lineplot(x=monthly_temp.index, y=monthly_temp.values, marker='o')
plt.title("Monthly Average Temperature (°C)")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.xticks(range(1, 13))
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Temp_C', y='Rel Hum_%', alpha=0.5, color='green')
plt.title("Temperature vs Relative Humidity")
plt.xlabel("Temperature (°C)")
plt.ylabel("Relative Humidity (%)")
plt.show()
```

```python
top_conditions = df['Weather'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(y=top_conditions.index, x=top_conditions.values, palette="viridis")
plt.title("Top 10 Most Frequent Weather Conditions")
plt.xlabel("Frequency")
plt.ylabel("Condition")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Month', y='Temp_C', palette="Set2")
plt.title("Monthly Temperature Distribution")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
top5_weather = df['Weather'].value_counts().head(5)

plt.figure(figsize=(7,7))
plt.pie(top5_weather.values, labels=top5_weather.index, autopct='%1.1f%%', startangle=140)
plt.title("Top 5 Weather Condition Distribution")
plt.axis('equal')
plt.show()
```

```python
hourly_avg = df.groupby('Hour')['Temp_C'].mean()

plt.figure(figsize=(10,5))
plt.fill_between(hourly_avg.index, hourly_avg.values, color='lightblue')
plt.plot(hourly_avg.index, hourly_avg.values, marker='o', color='blue')
plt.title("Average Temperature by Hour")
plt.xlabel("Hour")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,6))
corr = df[['Temp_C', 'Dew Point Temp_C', 'Rel Hum_%', 'Wind Speed_km/h', 'Visibility_km', 'Press_kPa']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
print("🔹 Summary of Weather Data Insights:")
print("- Avg Temp:", round(df['Temp_C'].mean(), 2), "°C")
print("- Max Temp:", df['Temp_C'].max(), "°C")
print("- Min Temp:", df['Temp_C'].min(), "°C")
print("- Most Frequent Condition:", df['Weather'].mode()[0])
print("- Coldest Hour:", df[df['Temp_C'] == df['Temp_C'].min()]['Date/Time'].values[0])
print("- Hottest Hour:", df[df['Temp_C'] == df['Temp_C'].max()]['Date/Time'].values[0])
```
